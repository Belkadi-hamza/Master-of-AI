# Séparer stats domicile / extérieur

```python
home_stats = teamstats[teamstats['location']=='h'].copy()
away_stats = teamstats[teamstats['location']=='a'].copy()
```

```python
home_stats.head()
```

```python
away_stats.head()
```

# Renommer colonnes

```python
home_stats = home_stats.add_prefix('home_')
away_stats = away_stats.add_prefix('away_')
```

```python
home_stats.head()
```

```python
away_stats.head()
```

# Fusion avec games

```python
df = games.merge(home_stats, left_on=['gameID','homeTeamID'], right_on=['home_gameID','home_teamID'])
df = df.merge(away_stats, left_on=['gameID','awayTeamID'], right_on=['away_gameID','away_teamID'])
```

```python
df.head()
```

# Définir target

```python
def match_result(row):
    if row['homeGoals'] > row['awayGoals']:
        return 'Home Win'
    elif row['homeGoals'] < row['awayGoals']:
        return 'Away Win'
    else:
        return 'Draw'
```

```python
df['target'] = df.apply(match_result, axis=1)
```

```python
df.head()
```

# Colonnes finales

```python
features = [
    'season','leagueID',
    'home_goals','home_xGoals','home_shots','home_shotsOnTarget','home_corners','home_ppda','home_yellowCards','home_redCards',
    'away_goals','away_xGoals','away_shots','away_shotsOnTarget','away_corners','away_ppda','away_yellowCards','away_redCards',
    'target'
]
```

```python
final_dataset = df[features]
```

```python
final_dataset.head()
```

```python
final_dataset['target_code'] = df['target'].map({'Home Win':0,'Draw':1,'Away Win':2})
```

```python
final_dataset.to_csv(folder_path+"/Data/dataset_match_classification.csv", index=False)
```
