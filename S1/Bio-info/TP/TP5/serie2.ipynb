## Exercice 2:

```python
def Intersection(T1, T2):
    T = []
    if len(T1)>0 and len(T2)>0:
        for t1 in T1:
            if t1 in T2:
                T.append(t1)
    return T
```

```python
T1=[0, 3 , 6, 8, 26, 19,33, 5, 11]
T2=[0, 3 , 8, 16, 12,33]
Intersection(T1,T2)
```

```python
def Complement(T):
    C = []
    if len(T)>0:
        max = T[-1]
    else:
        max = 0
    for e in range(max):
        if e not in T:
            C.append(e)
    return C
```

```python
T = [0, 1, 2, 3, 6, 8, 9, 11, 14, 15, 17, 18]
Complement(T)
```

```python
def Fusion(T1,T2):
    T3 = []
    if len(T1)==0:
        T3 = T1
    elif len(T2)==0:
        T3 = T2
    else:
        T3 = list(set(T1 + T2))
        T3.sort()
    return T3
```

```python
def FusionV2(T1,T2):
    for e in T1:
        if e not in T2:
            T2.append(e)
    T2.sort()
    return T2
```

```python
T1=[3 , 6, 8, 26, 109,133, 165, 181]
T2=[3 , 8, 109, 112,303]
print(Fusion(T1,T2))
print(FusionV2(T1,T2))
```

## Exercice 3:

```python
def frequence(T):
    F = []
    for e in set(T):
        F.append((e,T.count(e)))
    return F
```

```python
T = [9, 4, 5, 6, 7, 8, 1, 2, 6, 7, 8, 8, 10, 4, 6, 0, 1, 2, 3, 4, 5]
frequence(T)
```

## Exercice 4: Compression de donnés

```python
def fibo(n):
    if n == 0 or n == 1:
        return 1
    else:
        return fibo(n-1) + fibo(n-2)
```

```python
def fibo_iter(n):
    a, b = 1, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b
```

```python
n = 10
print(fibo_iter(n))
print(fibo(n))
```

```python
def SuiteFibo(n):
    L = [[],[]]
    for i in range(1,n+1):
        if fibo_iter(i) <= n:
            L[0].append(fibo_iter(i))
            L[1].append(0)
        else:
            break
    return L
```

```python
SuiteFibo(50)
```

```python
def coder(n):
    L = SuiteFibo(n)
    r = n
    L[0].reverse()
    for l in L[0]:
        if r == 0: break
        if (r-l) >= 0:
            index = L[0].index(l)
            L[1][index] = 1
            r -= l
    L[0].reverse()
    L[1].reverse()
    return L
    
```

```python
coder(50)
```

```python
def decoder(X):
    l = []
    for i in range(1,len(X)+1):
        l.append(fibo_iter(i))
    L = list(map(lambda a,b: a*b, X, l))
    return sum(L)
    
```

```python
X = [0, 0, 1, 0, 0, 1, 0, 1]
decoder(X)
```
