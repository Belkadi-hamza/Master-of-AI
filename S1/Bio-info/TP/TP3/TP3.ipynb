```python
def needleman_wunsch(sequence1,sequence2, match=1, mismatch=0, gap=-1):
    n = len(sequence1)
    m = len(sequence2)

    score = np.zeros((n+1, m+1))
    
    for i in range(n+1):
        score[i][0] = gap * i
    for j in range(m+1):
        score[0][j] = gap * j
    
    for i in range(1, n+1):
        for j in range(1, m+1):
            if sequence1[i-1] == sequence2[j-1]:
                diag = score[i-1][j-1] + match
            else:
                diag = score[i-1][j-1] + mismatch
            up = score[i-1][j] + gap
            left = score[i][j-1] + gap
            score[i][j] = max(diag, up, left)
```

```python
sequences_n = {
    sequences[0].id:[],
    sequences[1].id:[]
}
for i in sequences:
    for j in str(i.seq):
        if j == 'A':
            sequences_n[i.id].append(1)
        elif j == 'T':
            sequences_n[i.id].append(2)
        elif j == 'G':
            sequences_n[i.id].append(3)
        elif j == 'C':
            sequences_n[i.id].append(4)


sequences_n
```

```python
n = len(sequences[0].seq)
m = len(sequences[1].seq)

score = np.zeros((n+2, m+2))
for i in range(2, n+2):
    score[0][i] = sequences_n[sequences[0].id][i-2]
for j in range(2, m+2):
    score[j][0] = sequences_n[sequences[1].id][j-2]

gap = -1
for i in range(1, n+1):
    score[1][i] = i * gap
for j in range(1, m+1):
    score[j][1] = j * gap
score
```
