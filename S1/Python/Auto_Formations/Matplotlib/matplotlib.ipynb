```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]
z = [i*2 for i in y]
```

```python
plt.plot(x,y, color='red', label='Line 1')
plt.plot(x,z, color='blue', label='Line 2')
plt.legend()
plt.show()
```

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]
plt.scatter(x,y)
plt.show()
```

```python
x = ['A','B','C','D','E']
y = [1,4,9,16,25]
plt.bar(x,y)
plt.show()
```
