# Read Files

```python
workers_file = open("workers.txt",'r+')
if workers_file.readable():
    # print(workers_file.read()) # Read entire file content
    print(workers_file.readline()) # Read first line
    print(workers_file.readline()) # Read second line
    # ...
    print(workers_file.readlines()) # Read all lines into a list
workers_file.close()
```

# Write Files

```python
students_file = open("students.html",'w')
students_file.write("<h1>Students List</h1>\n")
ul = ["<ul>\n", "  <li>Alice</li>\n", "  <li>Bob</li>\n", "  <li>Charlie</li>\n", "</ul>\n"]
students_file.writelines(ul)
students_file.close()
```
