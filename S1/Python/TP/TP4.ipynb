```python
class Point:
    NbPoints = 0
    def __init__(self, x, y):
        self.__x = x
        self.__y = y
        Point.NbPoints += 1
    
    def deplace(self, dx, dy):
        self.__x += dx
        self.__y += dy
    
    def affiche(self):
        print(f"Point({self.__x}, {self.__y})")

    def homothetie(self, m):
        self.__x *= m
        self.__y *= m
    
    def PointSymmetriqueParRapportAO(self):
        return Point(-self.__x, -self.__y)
    
    def PointSymmetriqueParRapportAxeDesAbscisses(self):
        return Point(-self.__x, self.__y)
    
    def __str__(self):
        return f"Point({self.__x}, {self.__y})"
```

```python
P1 = Point(2,2)
P2 = P1.PointSymmetriqueParRapportAO()
P3 = P1.PointSymmetriqueParRapportAxeDesAbscisses()
print(P1)
print(P2)
print(P3)
```

```python
class Fraction:
    num  = 0
    den = 0
    def __init__(self, n, d):
        self.num = 0
        self.den = -1
        if n>=0:
            raise ValueError("Le numérateur m'ai pas positive")
        if d == 0:
            raise ValueError("Le dénominateur n'est pas nul")
        self.num = n
        self.den = d

    def __str__(self):
        if self.den == 1:
            return f"{self.num}"
        return f"{self.num/self.den}"
    
    def __eq__(self, other):
        return self.__str__() == other.__str__()
    
    def _pgcd(self, a, b):
        for i in range(min(abs(a),abs(b)),0,-1):
            if (a%i == 0) and (b%i == 0):
                return i
        return 1
    
    def _simplifie_en_place(self):
        pgcd = self._pgcd(self.num,self.den)
        self.num = int(self.num/pgcd)
        self.den = int(self.den/pgcd)

    def simplifie(self):
        pgcd = self._pgcd(self.num,self.den)
        return f"{int(self.num / pgcd)}/{int(self.den / pgcd)}"
        
    def __add__(self,other):
        n = self.num * other.den + other.num * self.den
        d = self.den * other.den
        return Fraction(n,d)
    
    def __sub__(self,other):
        n = self.num * other.den - other.num * self.den
        d = self.den * other.den
        return Fraction(n,d)
    
    def __mul__(self,other):
        n = self.num * other.num
        d = self.den * other.den
        return Fraction(n,d)
    
    def __truediv__(self,other):
        if other.num == 0:
            raise ZeroDivisionError("The num is equal zero and division by zero is imposible")
        n = self.num * other.den
        d = self.den * other.num
        return Fraction(n,d)
    
    def to_float(self):
        return float(self.__str__())
    
    def __lt__(self, other):
        return self.to_float() < other.to_float()
    
    def __le__(self, other):
        return self.to_float() <= other.to_float()
    
    def __gt__(self, other):
        return self.to_float() > other.to_float()
    
    def __ge__(self, other):
        return self.to_float() >= other.to_float()
    
```

```python
import numpy as np
class Etudiant:
    __slots__ = ('__FN','__LN','__NPY','__NBD','__NML','__NSA')

    def __init__(self, FirstName, LastName, N_Python, N_BigData, N_MachineLearning, N_statAvan):
        self.__FN = FirstName
        self.__LN = LastName
        if (N_Python <0 or N_Python>20) or (N_BigData <0 or N_BigData>20) or (N_MachineLearning <0 or N_MachineLearning>20) or (N_statAvan <0 or N_statAvan>20):
            raise ValueError("La note n'est pas entre 0 et 20. L'intervale valide est [0-20]")
        self.__NPY = N_Python
        self.__NBD = N_BigData
        self.__NML = N_MachineLearning
        self.__NSA = N_statAvan

    def __str__(self):
        return f"My name is {self.__FN} {self.__LN} my notes are (\n\tPython : {self.__NPY}\n\tBig Data : {self.__NBD}\n\tMachine Learning : {self.__NML}\n\tStat Avan : {self.__NSA}\n)"
    
    def moyenne(self):
        Mean = np.mean([self.__NPY,self.__NBD,self.__NML,self.__NSA])
        return Mean
    
    def mention(self):
        mean = self.moyenne()
        if mean>=16:
            return "Excellent"
        elif mean>=14:
            return "Bien"
        elif mean>=12:
            return "Assez bien"
        elif mean>=10:
            return "Passable"
        else:
            return "Insuffisant"
    
    def __lt__(self, other):
        return self.moyenne() < other.moyenne()
```

```python
class Personne:
    __slots__ = ('nom','prenom')
    def __init__(self):
        self.nom = "anonyme"
        self.prenom = "anonyme"

    def __str__(self):
        return f"Nom : {self.nom} , Prénom : {self.prenom}"
    
    def getNom(self):
        return self.nom
    
    def setNom(self,nom):
        self.nom = nom
    
    def getPrenom(self):
        return self.prenom
    
    def setPrenom(self,prenom):
        self.prenom = prenom
    
    def affiche(self):
        return f"je suis {self.getPrenom()} {self.getNom()}"
```

```python
class Etudiant(Personne):
    __slots__ = ('CNE', 'filiere')
    def __init__(self):
        self.CNE = "anonyme"
        self.filiere = "anonyme"
    
    def __str__(self):
        return f"{super().__str__()},  CNE : {self.CNE} , Filière : {self.filiere}"
    
    def getCNE(self):
        return self.CNE
    
    def setCNE(self,CNE):
        self.CNE = CNE
    
    def getFiliere(self):
        return self.filiere
    
    def setFiliere(self, filiere):
        self.filiere = filiere

    def  affiche(self):
        return f"{super().affiche()}, mon CNE est {self.getCNE()} et ma filiere {self.getFiliere()}"
```

```python
class CandidatPourUnConcours(Etudiant):
    __slots__ = ('numeroExam', 'moyenneGenerale')
    def __init__(self):
        self.numeroExam = 'anonyme'
        self.moyenneGenerale = 'anonyme'
    
    def __str__(self):
        return f"{super().__str__()}, Numéro d'examen : {self.numeroExam}, Moyenne Générale : {self.moyenneGenerale}"
    
    def getNumeroExam(self):
        return self.numeroExam
    
    def setNumeroExam(self,numeroExam):
        self.numeroExam = numeroExam
    
    def getMoyenneGenerale(self):
        return self.moyenneGenerale
    
    def setMoyenneGenerale(self,moyenneGenerale):
        self.moyenneGenerale = moyenneGenerale

    
    def affiche(self):
        return f"{super().affiche()}. Mon Numero d'examen est {self.getNumeroExam()} et Moyenne generale {self.getMoyenneGenerale()}"
```

```python
def afficheSelonType(personne):
    return personne.affiche()
```

```python
# une Personne 'HASSANI', 'Ismail' 
personne = Personne()
personne.setPrenom('HASSANI')
personne.setNom('Ismail')
# un Etudiant 'ALAMI', 'Ghita', 5678, 'Analyse Mathématiques'
etudiant = Etudiant()
etudiant.setPrenom('ALAMI')
etudiant.setNom('Ghita')
etudiant.setCNE('5678')
etudiant.setFiliere('Analyse Mathématiques')
# un CandidatPourUnConcours 'MOUMNI', 'Latifa', 3456, 'SMA', 5, 14.5 
CpC = CandidatPourUnConcours()
CpC.setPrenom('MOUMNI')
CpC.setNom('Latifa')
CpC.setCNE('3456')
CpC.setFiliere('SMA')
CpC.setNumeroExam(5)
CpC.setMoyenneGenerale(14.5)
```

```python
print(afficheSelonType(personne))
print(afficheSelonType(etudiant))
print(afficheSelonType(CpC))
```
