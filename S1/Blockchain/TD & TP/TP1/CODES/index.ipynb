```python
def cesar_crypt(message,cle):
    uppercase = (65,90)
    lowercase = (97,122)
    if (not -26<=cle<=26) or (len(message)<1):
        return "Cle is incorect or your message is empty"
    else:
        crypted_message = ''
        for char in message:
            position = ord(char)
            if char.isalpha():
                position += cle
                if char.isupper():
                    if position < uppercase[0]:
                        position += 26
                    elif position > uppercase[1]:
                        position -= 26
                elif char.islower():
                    if position < lowercase[0]:
                        position += 26
                    elif position > lowercase[1]:
                        position -= 26
            crypted_message += chr(position)
    return crypted_message
```

```python
def test_cesar_crypt_cle(message):
    for i in range(26):
        print(f"Cle : {i:02d} - Message : {cesar_crypt(message,i)}")

message = "ZHOFRPH WR WKH ZRUOG RI FUBSWRJUDSKB"    
test_cesar_crypt_cle(message)
```
