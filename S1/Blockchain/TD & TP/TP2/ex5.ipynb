```python
from Crypto.PublicKey import RSA
from Crypto.Cipher import PKCS1_OAEP
from Crypto.Hash import SHA256
from Crypto.Signature import pkcs1_15
```

1. Generer une de cles RSA avec une taille de 2048 bits

```python
key = RSA.generate(2048)
```

2. Exporter la cle privee et la cle publique au format PEM

```python
private_key = key.export_key()
with open("privateG1.pem", "wb") as f:
    f.write(private_key)
```

```python
public_key = key.publickey().export_key()
with open("publicG1.pem", "wb") as f:
    f.write(public_key)
```

3. Chiffrer un message avec la cle publique

```python
# Load the public key from the PEM file
with open("publicG1.pem", "rb") as f:
    public_key = RSA.import_key(f.read())
```

```python
# Encrypt the message
message = b"Hello, World!"
cipher = PKCS1_OAEP.new(public_key, hashAlgo=SHA256)
ciphertext = cipher.encrypt(message)
print("Message chiffré:", ciphertext)
```

4. Dechiffrer le message avec la cle privee

```python
cipher = PKCS1_OAEP.new(key, hashAlgo=SHA256)
plaintext = cipher.decrypt(ciphertext)
print("Message déchiffré:", plaintext)
```

5. Signer un message avec la cle privee

```python
signature = pkcs1_15.new(key).sign(SHA256.new(message))
print("Signature:", signature)
```

6. Verifier la signature avec la cle publique

```python
try:
    pkcs1_15.new(key.publickey()).verify(SHA256.new(message), signature)
    print("La signature est valide.")
except (ValueError, TypeError):
    print("La signature n'est pas valide.")
```
