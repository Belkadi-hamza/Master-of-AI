# 📝 Formulaire de Soumission de TP

Remplissez le formulaire ci-dessous pour envoyer votre travail pratique.

```python
# Installation des dépendances nécessaires
import sys
!{sys.executable} -m pip install -q python-dotenv ipywidgets

import smtplib
from email.message import EmailMessage
import os
from dotenv import load_dotenv
from datetime import datetime
import ipywidgets as widgets
from IPython.display import display, clear_output

# Charger les variables d'environnement
load_dotenv()

email_sender = os.getenv("EMAIL_SENDER")
email_password = os.getenv("EMAIL_PASSWORD")
email_receiver = os.getenv("EMAIL_RECEIVER")

# Nom et prénom fixes
NOM_FIXE = "Belkadi"
PRENOM_FIXE = "Hamza"

# Fonction d'envoi d'email
def envoyer_email(tp_type):
    try:
        date_envoi = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        
        msg = EmailMessage()
        msg['Subject'] = f"TP Soumission - {tp_type}"
        msg['From'] = email_sender
        msg['To'] = email_receiver
        
        msg.set_content(
            f"Bonjour Monsieur,\n\n"
            f"Veuillez trouver ci-joint mon Travail Pratique (TP).\n\n"
            f"Informations de l'étudiant :\n"
            f"Nom : {NOM_FIXE}\n"
            f"Prénom : {PRENOM_FIXE}\n"
            f"Type de TP : {tp_type}\n\n"
            f"Date d'envoi : {date_envoi}\n\n"
            f"Cordialement."
        )
        
        with smtplib.SMTP_SSL("smtp.gmail.com", 465) as smtp:
            smtp.login(email_sender, email_password)
            smtp.send_message(msg)
        
        return True, "Email envoyé avec succès!"
    
    except Exception as e:
        return False, f"Erreur lors de l'envoi: {str(e)}"


print("\n🎓 FORMULAIRE DE SOUMISSION DE TP\n" + "="*50 + "\n")
print(f"👤 Étudiant : {NOM_FIXE} {PRENOM_FIXE}\n")

# Liste des TP
tp_dropdown = widgets.Dropdown(
    options=[
        'correction_exercice_1',
        'correction_exercice_2',
        'correction_exercice_3',
        'correction_exercice_4',
        'correction_exercice_5',
        'correction_exercice_6',
        'correction_exercice_7'
    ],
    description='TP:',
    style={'description_width': '80px'},
    layout=widgets.Layout(width='400px')
)

# Bouton d'envoi
envoyer_btn = widgets.Button(
    description='📤 Envoyer',
    button_style='success',
    tooltip='Cliquez pour envoyer votre TP',
    layout=widgets.Layout(width='200px', height='40px')
)

# Message de statut
output_message = widgets.Output()

# Fonction de clic
def on_envoyer_click(b):
    with output_message:
        clear_output()
        
        if not tp_dropdown.value:
            print("❌ Erreur: Veuillez sélectionner un TP!")
            return
        
        succes, message = envoyer_email(tp_dropdown.value)
        
        if succes:
            print(f"✅ {message}")
            print("\n📋 Résumé de votre soumission:")
            print(f"   Nom: {NOM_FIXE}")
            print(f"   Prénom: {PRENOM_FIXE}")
            print(f"   TP: {tp_dropdown.value}")
            print(f"   Date/Heure: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
        else:
            print(f"❌ {message}")

envoyer_btn.on_click(on_envoyer_click)

# Afficher le formulaire
form = widgets.VBox([
    tp_dropdown,
    widgets.HTML("<br>"),
    envoyer_btn,
    output_message
])

display(form)
```
