## 🔹 Rappels mathématiques

### 1. Matrice symétrique
Une matrice $A$ est symétrique si :
$$ A^T = A $$

### 2. Déterminant
Le déterminant d'une matrice permet de savoir si elle est inversible.

### 3. Valeurs propres
Les valeurs propres $\lambda$ sont solutions de :
$$ \det(A - \lambda I) = 0 $$

### 4. Matrice définie positive
Une matrice est définie positive si :
- Toutes ses valeurs propres sont strictement positives
- Elle admet une décomposition de Cholesky

### 5. Décomposition de Cholesky
Une matrice définie positive peut s’écrire :
$$ A = LL^T $$
où $L$ est triangulaire inférieure.


```python
import sympy as sp
```

## 🔹 1) Définir la matrice A

```python
# ==========================
# 📌 Définition de la matrice
# ==========================
# TODO : définir la matrice A (3x3)
A = sp.Matrix([[2,-1,0],[-1,2,-1],[0,-1,2]])
```

## 🔹 2) Afficher la matrice

```python
print(" A :")
# TODO : Afficher la matrice A (3x3)
sp.pprint(A)
```

## 🔹 3) Calcul du déterminant

```python

# TODO : Calculer le déterminant de A  
det_A= A.det()
print("Déterminant de A :", det_A)
```

## 🔹 4) Calcul de la transposée

```python

# TODO : Calculer le transposée de A  

print("Transposée de A :")
A_T = A.T
sp.pprint(A_T)
```

## 🔹 5) Valeurs propres

```python
# TODO : Calculer les valeurs propres de A  
eigenvalues= A.eigenvals()
print("Valeurs propres de A :")
sp.pprint(eigenvalues)
```


## 🔹 6) Vérification (valeurs propres)

Une matrice est définie positive si toutes ses valeurs propres sont positives.


```python
# test de positivité
# TODO : vérifier si toutes les valeurs propres sont positives
if all(val > 0 for val in eigenvalues.keys()):
    print("Toutes les valeurs propres sont positives.")
else:
    print("Certaines valeurs propres ne sont pas positives.")
```


## 🔹 7) Vérification (Cholesky)

Si la décomposition de Cholesky échoue → la matrice n’est pas définie positive.


```python
# TODO : Completé le code
try:
    L = A.cholesky()
    print("Décomposition de Cholesky réussie :")
    sp.pprint(L)
    print("Donc A est définie positive")
except ValueError:
    print("Échec de la décomposition de Cholesky")
    print("Donc A n'est PAS définie positive")
```

Methode 2

```python
# valeurs propres positives et A symetrique
# => Décomposition de Cholesky réussie alor A est definie positive
if all(val > 0 for val in eigenvalues.keys()) and A == A.T:
    print("Décomposition de Cholesky réussie")
else:
    print("Décomposition de Cholesky échouée")
```
