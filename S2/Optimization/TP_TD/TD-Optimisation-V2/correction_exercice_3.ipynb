# Exercice 3: Matrices, valeurs propres et vecteurs propres

## 1. Définition des matrices
Rappel : Une matrice carrée $A$ admet des **valeurs propres** $\lambda$ telles que :
$A v = \lambda v$ où $v$ est un vecteur propre.

```python
# ==========================
# 🧮 Exercice SymPy - Matrices
# ==========================

import sympy as sp
from IPython.display import display, Math
```

```python
# ==========================
# 📌 Définition des matrices
# ==========================
# TODO : définir la matrice A
A = sp.Matrix([[2,1],[1,2]])

# TODO : définir la matrice B
B = sp.Matrix([[3,1],[5,10]])
```

```python
# ==========================
# 📌 Affichage des matrices
# ==========================
# TODO : afficher A en LaTeX
display(Math(r"A = " + sp.latex(A)))

# TODO : afficher B en LaTeX
display(Math(r"B = " + sp.latex(B)))
```

## 2. Affichage des matrices

```python
# ==========================
# 📌 Affichage des matrices
# ==========================
# TODO : afficher A avec pprint (affichage propre)
print("Matrice A :")
sp.pprint(A)

# TODO : afficher B avec pprint (affichage propre)
print("\nMatrice B :")
sp.pprint(B)
```

## 3. Valeurs propres et vecteurs propres
Rappel :
- Les valeurs propres sont solutions de $\det(A - \lambda I)=0$
- Les vecteurs propres sont associés à chaque valeur propre.

## Calcul des valeurs propres

```python
# TODO : calculer les valeurs propres et vecteurs propres de A
eigvals_A = A.eigenvals()
eigvecs_A = A.eigenvects()
```

```python
eigvals_A
```

```python
eigvecs_A
```

## Affichage des résultats

```python
display(Math(r"\text{Valeurs propres et vecteurs propres de } A :"))


# TODO : parcourir eig_A et afficher les résultats
for val, mult, vects in eigvecs_A:

    # TODO : convertir la valeur propre en LaTeX
    latex_val = sp.latex(val)

    # TODO : convertir la multiplicité en LaTeX
    latex_mult = sp.latex(mult)

    # TODO : convertir les vecteurs propres en LaTeX
    latex_vecs = sp.latex(vects)

    # TODO : afficher le résultat formaté
    display(Math(
        rf"\lambda = {latex_val},\quad "
        rf"\text{{multiplicité }} {latex_mult},\quad "
        rf"\text{{vecteurs propres : }} {latex_vecs}"
    ))
```
